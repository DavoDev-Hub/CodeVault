# Importaciones

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Datos del tiempo

```python
tiempo = [2.1, 3.5, 
          1.8, 4.0, 2.9, 3.2, 2.7, 
          3.8, 2.4, 3.1, 2.2, 3.3, 
          1.9, 3.9, 2.8, 3.5, 2.6, 
          3.4, 2.5]
```

```python
media = np.mean(tiempo)
varianza = np.var(tiempo)
desviacionS = np.std(tiempo)
    

print("Media: ", media)
print("Varianza: ", varianza)
print("Desviación Estándar: ", desviacionS)
```

```python
plt.hist(tiempo, bins=4, edgecolor='black')

plt.title("Histograma de los tiempos")
plt.xlabel("Tiempo")
plt.ylabel("Frecuencia")

plt.show()
```
