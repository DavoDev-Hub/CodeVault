```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
llamadas =  [12, 15, 14, 13, 16, 18, 11, 14, 17, 15, 
            10, 13, 11, 12, 14, 13, 15, 12, 16, 11]
```

```python
media = np.mean(llamadas)
varianza = np.var(llamadas)
desviacionS = np.std(llamadas)  
    

print("Media: ", media)
print("Varianza: ", varianza)
print("Desviación Estándar: ", desviacionS)
```

```python
plt.hist(llamadas, bins=4, edgecolor='black')

plt.title("Histograma de los tiempos")
plt.xlabel("Tiempo")
plt.ylabel("Frecuencia")

plt.show()
```
